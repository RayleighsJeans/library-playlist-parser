```python
import json

# Load Wikipedia albums
with open(file='test/wikipedia_bestselling_albums.json', mode='r') as f:
    albums = json.load(f)

# Search for each album
searcher: fas.StreamingSearcher = fas.StreamingSearcher()

for album_data in albums[:3]:  # First 3 albums
    artist = album_data['artist']
    album = album_data['album']

    print(f"\n{artist} - {album}")
    results: dict[str, str | None] = searcher.search_all_services(artist, album)

    for service, url in results.items():
        if url:
            print(f"  ✓ {service}: {url}")
```
